```r
ggplot(march_tidy, aes(x = reg_wins_pct)) +
  geom_histogram(binwidth = 5) +
  labs(x = "Regular Season Win Percentage (%)",
       y = "Number of Teams",
       title = "Distribution of Regular Season Win Percentage") +
  theme(text = element_text(size = 12))
```

```r
finish_percentage <- march_tidy |>
  group_by(tourney_finish) |>
  summarize(percentage = n() / nrow(march_tidy) * 100)

ggplot(finish_percentage, aes(x = percentage, y = tourney_finish)) +
  geom_bar(stat = "identity") +
  labs(x = "Percentage of Teams (%)",
       y = "Tournament Finish",
       title = "Proportion of Teams by Tournament Finish") +
  theme(text = element_text(size = 12))
```

Visualization 1:This plot reveals the proportion of each tournament finish for every possible seed.
